# Exploring Class Sizes in NYC Public Schools

In this project, I will use data from NYC Open Data to explore class sizes in New York City public schools during the 2021–2022 school year. I will calculate the mean, median, and mode of average class sizes to better understand how class sizes vary across the dataset.

## Dataset

The dataset comes from NYC Open Data and contains information about class sizes in NYC public schools during the 2021–2022 school year.

Source: [2021–2022 Average Class Size by School](https://data.cityofnewyork.us/Education/2021-2022-Average-Class-Size-by-School/sgr7-hhwp)

```python
import pandas as pd

df = pd.read_csv("2021_-_2022_Average_Class_Size_by_School_20261113.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
class_sizes = df["Average Class Size"]
```

```python
mean = class_sizes.mean()
print(mean)
```

```python
median = class_sizes.median()
print(median)
```

```python
mode = class_sizes.mode()
print(mode)
```

## Results

Using pandas, I calculated the mean, median, and mode of average class sizes.

The mean is 20.84, while both the median and mode are 22. This shows that the mean is slightly lower than the median, and 22 is the most common value in the dataset.

## Analyzing the Data Without Pandas

```python
import csv
```

```python
class_sizes_list = []

with open("2021_-_2022_Average_Class_Size_by_School_20261113.csv", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        size = float(row["Average Class Size"])
        class_sizes_list.append(size)

print(len(class_sizes_list))
```

```python
total = 0

for size in class_sizes_list:
    total += size

mean1 = total / len(class_sizes_list)

print(mean1)
```

```python
sorted_sizes = sorted(class_sizes_list)
```

```python
middle = len(sorted_sizes) // 2
median1 = sorted_sizes[middle]
median1
```

```python
counts = {}

for size in class_sizes_list:
    if size in counts:
        counts[size] += 1
    else:
        counts[size] = 1
```

```python
max_count = 0
mode1 = None

for size in counts:
    if counts[size] > max_count:
        max_count = counts[size]
        mode1 = size

mode1
```


I calculated the mean, median, and mode using both pandas and basic Python.

Both methods gave the same results: a mean of about 20.84, a median of 22, and a mode of 22.0. This confirms that my manual calculations match the pandas results.

## Visualizing Class Sizes

```python
groups = [0, 0, 0, 0, 0]

for size in class_sizes_list:
    if size < 10:
        groups[0] += 1
    elif size < 20:
        groups[1] += 1
    elif size < 30:
        groups[2] += 1
    elif size < 40:
        groups[3] += 1
    else:
        groups[4] += 1

groups
```

```python
labels = ["0-9", "10-19", "20-29", "30-39", "40+"]

print("NYC Average Class Size Distribution")
print("Each * represents approximately 250 records")

for i in range(len(groups)):
    stars = round(groups[i] / 250)
    print(labels[i] + ": " + "*" * stars, groups[i])
```

I used one star to represent approximately 250 records so that the chart would be easier to read.

The chart shows that most records have an average class size between 20 and 29 students. There are 7,515 records in this group, while only 6 records have an average class size of 40 or more students.
