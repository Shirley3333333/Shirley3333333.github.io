# NYC Taxi Trip Data Analysis

```python
import csv

with open("nyc_data.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["trip_distance"])
```

```python
distances = []

with open("nyc_data.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        if row["trip_distance"]:
            distance = float(row["trip_distance"])
            distances.append(distance)

average = sum(distances) / len(distances)

print("Average ride distance:", average)
```

```python
total_trips = 0
credit_card_trips = 0

with open("nyc_data.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        total_trips += 1

        if row["payment_type"] == "1":
            credit_card_trips += 1

percentage = credit_card_trips / total_trips * 100

print("Credit card percentage:", percentage)
```

```python
import random

with open("nyc_data.csv", "r") as file:
    reader = csv.DictReader(file)
    trips = list(reader)
    fieldnames = reader.fieldnames

sample = random.sample(trips, 100)

with open("sample_trips.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(sample)

print("Sample saved successfully!")
```
